```python
## 特征工程部分增强
feature_cols = [col for col in numerical_cols if col not in ['SaleID','name','regDate','creatDate','price','model','brand','regionCode','seller']]
feature_cols = [col for col in feature_cols if 'Type' not in col]

# 填充缺失值
X_data = Train_data[feature_cols].fillna(-1)
Y_data = Train_data['price']
X_test = TestB_data[feature_cols].fillna(-1)

# 确保 'power' 和 'kilometer' 列为数值类型
Train_data['power'] = pd.to_numeric(Train_data['power'], errors='coerce').fillna(-1)
Train_data['kilometer'] = pd.to_numeric(Train_data['kilometer'], errors='coerce').fillna(-1)
TestB_data['power'] = pd.to_numeric(TestB_data['power'], errors='coerce').fillna(-1)
TestB_data['kilometer'] = pd.to_numeric(TestB_data['kilometer'], errors='coerce').fillna(-1)

# 添加高级特征并处理无穷大值
X_data['power_per_kilometer'] = Train_data['power'] / np.maximum(Train_data['kilometer'], 1)
X_data['age'] = 2020 - Train_data['regDate'].astype(int).astype(str).str[:4].astype(int)
X_data['mile_per_year'] = Train_data['kilometer'] / np.maximum(X_data['age'], 1)
X_data['power_age_ratio'] = Train_data['power'] / np.maximum(X_data['age'], 1)

X_test['power_per_kilometer'] = TestB_data['power'] / np.maximum(TestB_data['kilometer'], 1)
X_test['age'] = 2020 - TestB_data['regDate'].astype(int).astype(str).str[:4].astype(int)
X_test['mile_per_year'] = TestB_data['kilometer'] / np.maximum(X_test['age'], 1)
X_test['power_age_ratio'] = TestB_data['power'] / np.maximum(X_test['age'], 1)

# 处理无穷大和异常值
X_data = X_data.replace([np.inf, -np.inf], np.nan).fillna(-1)
X_test = X_test.replace([np.inf, -np.inf], np.nan).fillna(-1)

# 添加多项式特征
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2, interaction_only=True, include_bias=False)
X_data_poly = poly.fit_transform(X_data)
X_test_poly = poly.transform(X_test)

## 更新特征列
original_features = X_data.columns.tolist()
poly = PolynomialFeatures(degree=2, interaction_only=True, include_bias=False)
X_data_poly = poly.fit_transform(X_data)
X_test_poly = poly.transform(X_test)

feature_cols_poly = poly.get_feature_names_out(original_features)
X_data = pd.DataFrame(X_data_poly, columns=feature_cols_poly)
X_test = pd.DataFrame(X_test_poly, columns=feature_cols_poly)

print('X train shape:',X_data.shape)
print('X test shape:',X_test.shape)
```

```python
from sklearn.model_selection import KFold

# 特征工程：标准化
scaler = preprocessing.StandardScaler()
X_data_scaled = scaler.fit_transform(X_data)
X_test_scaled = scaler.transform(X_test)

## 优化XGBoost模型参数
def build_model_xgb(x_train, y_train):
    model = xgb.XGBRegressor(
        n_estimators=1000,
        max_depth=7,
        learning_rate=0.008,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.1,
        reg_lambda=0.1,
        min_child_weight=1,
        gamma=0.1,
        tree_method='gpu_hist',
        random_state=42
    )
    model.fit(
        x_train, 
        y_train,
        eval_metric='mae',
        early_stopping_rounds=100,
        verbose=False
    )
    return model

## 优化LightGBM模型参数
def build_model_lgb(x_train, y_train):
    estimator = lgb.LGBMRegressor(
        num_leaves=63,
        n_estimators=1000,
        max_depth=7,
        learning_rate=0.008,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.1,
        reg_lambda=0.1,
        min_child_weight=1,
        random_state=42
    )
    param_grid = {
        'learning_rate': [0.005, 0.008, 0.01],
        'n_estimators': [800, 1000, 1200]
    }
    gbm = GridSearchCV(estimator, param_grid, cv=5, scoring='neg_mean_absolute_error')
    gbm.fit(x_train, y_train)
    return gbm
```

```python
# 数据划分
# 按时间顺序划分（假设数据按时间排序）
split_index = int(len(X_data)*0.7)
x_train, x_val = X_data[:split_index], X_data[split_index:]
y_train, y_val = Y_data[:split_index], Y_data[split_index:]

# 修正后的XGBoost模型定义
def build_model_xgb(x_train, y_train):
    model = xgb.XGBRegressor(
        n_estimators=1000,
        max_depth=7,
        learning_rate=0.008,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.1,
        reg_lambda=0.1,
        min_child_weight=1,
        gamma=0.1,
        tree_method='gpu_hist',
        random_state=42,
        eval_metric='mae',          # 初始化时设置eval_metric
        early_stopping_rounds=100   # 初始化时设置early_stopping_rounds
    )
    model.fit(
        x_train, 
        y_train,
        eval_set=[(x_val, y_val)],  # 必须提供验证集
        verbose=False
    )
    return model

# 训练模型
print('Train xgb...')
model_xgb = build_model_xgb(x_train, y_train)
```

```python
## 改进模型融合策略
def ensemble_predict(lgb_model, xgb_model, x_val, weights=[0.7, 0.3]):
    lgb_pred = lgb_model.predict(x_val)
    xgb_pred = xgb_model.predict(x_val)
    ensemble = weights[0] * lgb_pred + weights[1] * xgb_pred
    # 后处理
    ensemble[ensemble < 0] = 10
    ensemble[ensemble > 200000] = 200000
    return ensemble

## 这里我们采取了简单的加权融合的方式
val_Weighted = ensemble_predict(model_lgb, model_xgb, x_val, weights=[0.7, 0.3])
print('MAE of val with Weighted ensemble:',mean_absolute_error(y_val,val_Weighted))
```

```python
# 修正模型变量名（根据实际训练时的变量名调整）
# 假设原训练代码中模型保存为 model_xgb 和 model_lgb
sub_Weighted = ensemble_predict(model_lgb, model_xgb, X_test, weights=[0.7, 0.3]) 

## 查看预测值的分布
plt.hist(sub_Weighted)  # 这里应该是预测值的直方图，原代码误用了Y_data
plt.show()
plt.close()
```
